```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
```

```python
D = 1.0 # [m^2/s]
L = 1.0 # [m]
ns = [2,4,6,8,10]
c = 1.0
t = 0.05 # [s]
eps_quad = 1e-3
eps_lchs = 1e-3
num_terms_list = []
for n in ns:
    N = 2**n
    norm = D * np.pi**2 * N**2 / L**2
    h = np.pi / (t*norm/2 + np.log(64*np.exp(3*c/2)/(15*eps_quad)))
    gamma = 1/c * np.sqrt(c+np.log((1+1/(2*np.pi))/eps_lchs))
    R = 2*c*gamma**2
    J = R/h
    print(f"J :{J}")
    J_int = int(2**(np.floor(np.log2(J)))) # closest power of 2 >= J
    print(f"J_int :{J_int}")
    R = J_int * h
    J = J_int
    # LCU terms
    num_terms = 2*J_int
    num_terms_list.append(num_terms)

plt.plot(ns, num_terms_list, marker='o')
plt.plot(ns, (2**np.array(ns))**2, marker='o')
```
